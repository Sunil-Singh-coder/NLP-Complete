**Word Lemmitaizer**
Lemmatization is the process of converting a word into its base dictionary form, called the **lemma**.
It uses vocabulary and grammar rules to find the correct root word.


```python
# Lemmitaization is better than Stemming 
from nltk.stem import WordNetLemmatizer

lema1= WordNetLemmatizer()   
'''POS 
Noun- n
Verb-v
adveb- r
adjective a
'''
#defalt it take pos='n" 
lema1.lemmatize("going")  # result is giving only going is pos ='n'
```

```python
lema2= WordNetLemmatizer()  

print(lema1.lemmatize("going",pos='v'))  # result is giving only go if pos ='n'
# Because going is verb 
print(lema1.lemmatize("history",pos='v'))  # result is giving only go if pos ='n'
print(lema1.lemmatize("studied",pos='v'))  # result is giving only go if pos ='n'
print(lema1.lemmatize("occuring",pos='v'))  # result is giving only go if pos ='n'

```

```python
words= ["playing","honesty","walking","farely","stydies","reading","history","swimming","singing","broken"]
for word in words:
    print(word + "---->" + lema2.lemmatize(word,pos='v'))
```
