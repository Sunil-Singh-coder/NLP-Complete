```python
para="""
Today I would like to talk about the importance of reading books. 
Reading is one of the best habits that a person can develop. 
It helps us improve our knowledge,vocabulary, and communication skills.
When we read books, we learn about different people, places, ideas, and cultures.
We can also improve our imagination and creativity. In today's world, many people
spend most of their time on social media, but they often forget the value of reading.
A good book can teach us important life lessons and help us make better decisions. 
Even if we read for only twenty minutes every day, we can become more confident and knowledgeable.
I believe that every student should develop the habit of reading because it improves thinking,
writing, and speaking skills. 
Let us make reading a part of our daily routine and continue learning throughout our lives. Thank you."""
```

```python
# import nltk
# nltk.download("stopwords")
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
stemming=PorterStemmer()
```

```python
stopwords.words("english")  #  here is the all stopwords that we have to removed 
```

```python
sentences=nltk.sent_tokenize(para)    # convert into paraghraph into list of sentences
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    # print("m")
    words=[stemming.stem(word) for word in words if word not in stopwords.words("english")]
    sentences[i]=' '.join(words)
    
print(sentences)    
    
    
```

```python
#  we use Word net lemmatizer to perform Stopwords  
from nltk.stem import WordNetLemmatizer
lemmitizer=WordNetLemmatizer()
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    # print("m")
    words=[lemmitizer.lemmatize(word.lower(),pos='v') for word in words if word not in stopwords.words("english")]
    sentences[i]=' '.join(words)
    
print(sentences)    
    
    
```
