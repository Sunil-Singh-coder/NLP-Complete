
****Stemming**** 
****Definition:****
*Stemming is the process of reducing a word to its root (stem) by removing prefixes or suffixes. The resulting stem may or may not be a valid English word.*




```python
from nltk.stem import PorterStemmer 
# example  
words= ["playing","honesty","walking","farely","stydies","reading","history","swimming","singing","broken"]
```

```python
1----> # It gives output not full accuracy so we it recover by lemmitaization 
stemming = PorterStemmer()
for word in words:
    print(word + "---->"+stemming.stem(word))
```

```python
# 2------># It support multiple lanugage
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english') # it support lanugage
for word in words:
    print(word + "---->"+snowball.stem(word))
```
