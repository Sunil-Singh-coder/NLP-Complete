```python
# from gensim.models import Word2Vec,KeyedVectors
from huggingface_hub import hf_hub_download
from gensim.models import KeyedVectors

# Download files from Hugging Face
kv_file = hf_hub_download(
    repo_id="Sunil-Singh-Coder/word2vec-google-news-300",
    filename="word2vec-google-news-300.kv"
)

vectors_file = hf_hub_download(
    repo_id="Sunil-Singh-Coder/word2vec-google-news-300",
    filename="word2vec-google-news-300.kv.vectors.npy"
)

# Load the model
wv = KeyedVectors.load(kv_file, mmap="r")


```

```python
# We download this model from hugging face
# This about how to download the model and how to use using word2Vec in our  Problem
import gensim.downloader as api
wv=api.load("word2vec-google-news-300")
```

```python
# now we can see the vector of any word which is coded in 300 dimensonal
wv["king"].shape
```

```python
# we can also able to find similar word by this gensim library

wv.most_similar("cricket")
wv.most_similar("happy")
```

```python
# we can also able to find what percetage has simlarity between voacab

wv.similarity("hockey","sports")     # 0.53541523
wv.similarity("happy","glad")        # 0.74088913
```

```python
# we also perform sub,add,mul so that we create different words
vec =wv["king"] * wv["man"]+wv["women"]
vec =wv["king"] + wv["man"]+wv["women"]
vec
wv.most_similar([vec])
# wv.most_similar("man")
```
